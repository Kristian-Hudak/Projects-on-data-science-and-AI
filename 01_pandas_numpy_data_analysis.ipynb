### Task 1: Total Coffee Sales Aggregation


Write a function `total_sold_by_type(coffee)` that calculates the total number of coffees sold for each coffee type over the whole week from the `coffee` DataFrame.  
The result must be a `pandas.Series` where the index is the coffee type (`typ_kavy`) and the values represent total units sold (`pocet_predanych`).


````python
def total_sold_by_type(coffee):
    """
    Summarizes the total number of sold coffee units grouped by type.
    """
    
    return(coffee.groupby(['typ_kavy'])['pocet_predanych'].sum())


````

### Task 2: Tallest Olympic Athlete Identification


Identify the tallest athlete in the `bios` dataset.  
The function should return a tuple in the format `(name, height)` representing the athlete's name and height in centimeters.


````python
def tallest_athlete(bios):
    """
    Returns the name and height (in cm) of the tallest athlete in the bios dataset.
    """


    zoradit=bios.sort_values(by="height_cm", ascending=False)
    zoradene=(zoradit.loc[:, ["name", "height_cm"]])


    prvy=zoradene.iloc[0]
    return (prvy["name"],int(float(prvy["height_cm"])))


    


````

### Task 3: Best Student Grade Average Calculation (NumPy)


Given a dictionary where each key is a student's name and the value is a NumPy array of grades:
```python
grades = {
    "Anna": np.array([1, 2, 1, 3]),
    "Peter": np.array([2, 2, 3, 1]),
    "Jana": np.array([1, 1, 1, 2])
}
```
Determine which student achieved the best (lowest numeric) average grade. The function should return the name of the top-performing student.


````python
grades = {
    "Anna": np.array([1, 2, 1, 3]),
    "Peter": np.array([2, 2, 3, 1]),
    "Jana": np.array([1, 1, 1, 2])
} # For this input, 'Jana' has the best average grade

def who_is_the_best(grades):

    priemeranna = np.mean(grades["Anna"])
    priemerpeter = np.mean(grades["Peter"])
    priemerjana = np.mean(grades["Jana"])
    if priemeranna < priemerpeter and priemeranna < priemerjana:
        return "Anna"
    elif priemerpeter < priemeranna and priemerpeter < priemerjana:
        return "Peter"
    else:
        return "Jana"

print(who_is_the_best(grades))
````
