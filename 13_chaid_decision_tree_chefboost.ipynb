This project implements non-binary decision tree regression using the **CHAID** (Chi-Square Automatic Interaction Detector) algorithm via `Chefboost`:
- Multi-way rule splitting based on categorical and continuous medical predictors (`age, sex, bmi, children, smoker, region`)
- Regression tree fitting on continuous medical insurance expenditure (`charges`)
- Graphviz visual tree rendering and rule extraction (`rules.py`)
- Model evaluation with regression metrics (MAE, RMSE, RAE, RRSE)


```python
import pandas as pd
df = pd.read_csv("/content/insurance.csv")

!pip install chefboost
from chefboost import Chefboost as chef

data = df.copy()

config = {"algorithm": "CHAID"}  # CHAID decision tree algorithm configuration

model = chef.fit(data, config, target_label="charges")
!pip install graphviz
from graphviz import Source
import os

dot_path = "outputs/tree.dot"

if os.path.exists(dot_path):
    with open(dot_path, "r") as f:
        dot_graph = f.read()
    print("=== CHAID Decision Tree Visualization ===")
    display(Source(dot_graph))
else:
    print("DOT graph file not found.")
rules_path = "outputs/rules/rules.py"

if os.path.exists(rules_path):
    with open(rules_path, "r") as f:
        rules_text = f.read()
    print("=== CHAID Decision Tree Generated Rules ===")
    print(rules_text)
else:
    print("rules.py rule file not found.")
```
